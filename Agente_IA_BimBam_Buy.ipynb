```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/perguntas_frequentes.pdf")
documentos = loader.load()

print(f"PDF carregado com {len(documentos)} páginas.")
print(documentos[0].page_content[:300])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documentos)

print(f"Documento dividido em {len(chunks)} chunks.")
print(chunks[0].page_content[:300])
```

```python
!pip install -U -q langchain-chroma
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="/content/chroma_db"
)

print(f"ChromaDB criado com sucesso! {vectorstore._collection.count()} documentos indexados.")
```

```python
pergunta = "Quais métodos de pagamento estão disponíveis?"

resultados = vectorstore.similarity_search(pergunta, k=3)

for i, doc in enumerate(resultados):
    print(f"--- Resultado {i+1} ---")
    print(doc.page_content[:300])
    print()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.3
)

resposta = llm.invoke("Diga apenas 'Conexão com o Gemini funcionando!' se você me entender.")
print(resposta.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
Você é um assistente que responde perguntas com base no contexto fornecido.
Use apenas as informações do contexto abaixo para responder. Se não souber a resposta
com base no contexto, diga que não encontrou essa informação no documento.

Contexto:
{context}

Pergunta: {input}

Resposta:
""")

print("Prompt criado com sucesso!")
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})

def formatar_contexto(documentos):
    return "\n\n".join(doc.page_content for doc in documentos)

print("Retriever configurado com sucesso!")
```

```python
def fazer_pergunta(pergunta):
    documentos_relevantes = retriever.invoke(pergunta)
    contexto = formatar_contexto(documentos_relevantes)

    prompt_formatado = prompt.format(context=contexto, input=pergunta)
    resposta = llm.invoke(prompt_formatado)

    # O Gemini 3.x pode retornar o conteúdo como lista de blocos (texto + metadados).
    # Este trecho extrai apenas o texto, seja qual for o formato retornado.
    if isinstance(resposta.content, str):
        texto_resposta = resposta.content
    else:
        texto_resposta = "".join(
            bloco.get("text", "")
            for bloco in resposta.content
            if isinstance(bloco, dict) and bloco.get("type") == "text"
        )

    return texto_resposta, documentos_relevantes

print("Função RAG corrigida com sucesso!")
```

```python
resposta, fontes = fazer_pergunta("Quais métodos de pagamento estão disponíveis?")

print("RESPOSTA:")
print(resposta)
print()
print("--- Fontes usadas ---")
for doc in fontes:
    print(doc.page_content[:150])
    print("---")
```

```python
def responder_chat(mensagem, historico):
    resposta, fontes = fazer_pergunta(mensagem)
    return resposta

print("Função de chat criada com sucesso!")
```

```python
import gradio as gr

interface = gr.ChatInterface(
    fn=responder_chat,
    title="Assistente BimBam Buy",
    description="Tire suas dúvidas sobre métodos de pagamento, reembolsos e mais, com base no documento oficial.",
    examples=[
        "Quais métodos de pagamento estão disponíveis?",
        "Como funciona o reembolso?",
        "É possível parcelar a compra?",
    ],
)

interface.launch(share=True)
```
